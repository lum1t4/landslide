```python
from itertools import product
from pathlib import Path

datas = [
    'data/processed/csv_split/A19_5cm.yml' ,
    'data/processed/csv_split/Avigliano_5cm.yml' ,
    'data/processed/csv_split/Avigliano2_5_cm.yml' ,
    'data/processed/csv_split/Bella_5cm.yml' ,
    'data/processed/csv_split/Bella3_5cm.yml' ,
    'data/processed/csv_split/Corleto_Perticara_5cm.yml' ,
    'data/processed/csv_split/Ferrandina2_5cm.yml' ,
    'data/processed/csv_split/Genzano_Lucania_5cm.yml' ,
    'data/processed/csv_split/Grassano_5cm.yml' ,
    'data/processed/csv_split/Ischia_5cm.yml' ,
    'data/processed/csv_split/Miglionico_5cm.yml' ,
    'data/processed/csv_split/Montescaglioso_5cm.yml' ,
    'data/processed/csv_split/Pietragalla_5cm.yml' ,
    'data/processed/csv_split/Pomarico_5cm.yml' ,
    'data/processed/csv_split/Provincia_Taranto_Ginosa_5cm.yml' ,
    'data/processed/csv_split/Provincia_Taranto_Montemesola_5cm_1.yml' ,
    'data/processed/csv_split/Provincia_Taranto_Montemesola_5cm_2.yml' ,
    'data/processed/csv_split/Provincia_Taranto_Montemesola_5cm_3.yml' ,
    'data/processed/csv_split/Stigliano_5cm.yml' ,
    'data/processed/csv_split/Stigliano2_5cm.yml' ,
    'data/processed/csv_split/Stigliano3_5cm.yml' ,
    'data/processed/csv_split/Stigliano4_5cm.yml',
]
losses = ["bce", "wbce", "focal_loss", "bce+dice", "bce+lovasz"]
models = ["L4S/unet"]


NUM_GPUS = 1
for i, (data, loss, model) in enumerate(product(datas, losses, models)):
    dix = i % NUM_GPUS

    image_sz = 512
    mask_sz = 128

    if "unet" in model: image_sz = mask_sz
    if "segformer" in model: mask_sz = image_sz // 4

    name = data.replace('data/processed/csv_split/', '').replace('.yml', '')
    name = f"{name}_{loss}_{model}"
    
    base = f"uv run scripts/recipe.py --config 'config/base.yml' --name '{name}' --model '{model}' --dataset '{data}' --criterion '{loss}' --device 'cuda:{dix}' --image_sz {image_sz} --mask_sz {mask_sz} --tracker 'wandb'" #  
    print(base)
```
